```python
senha = "123456"
senha_criptografada = ""

print("Senha original:", senha)
```

```python
for numero in senha:
    numero = int(numero) + 1
    senha_criptografada += str(numero)

print("Senha transformada:", senha_criptografada)
```

```python
senha_original = ""

for numero in senha_criptografada:
    numero = int(numero) - 1
    senha_original += str(numero)

print("Senha recuperada:", senha_original)
```

```python
def criptografar(mensagem):
    mensagem_criptografada = ""

    for caractere in mensagem:
        if caractere.isdigit():
            numero = int(caractere)
            novo_numero = (numero + 1) % 10
            mensagem_criptografada += str(novo_numero)
        else:
            mensagem_criptografada += caractere

    return mensagem_criptografada
```

```python
def descriptografar(mensagem):
    mensagem_original = ""

    for caractere in mensagem:
        if caractere.isdigit():
            numero = int(caractere)
            numero_original = (numero - 1) % 10
            mensagem_original += str(numero_original)
        else:
            mensagem_original += caractere

    return mensagem_original
```

```python
senha = input("Digite uma senha numérica para teste: ")

senha_codificada = criptografar(senha)
senha_recuperada = descriptografar(senha_codificada)

print("Senha original:", senha)
print("Senha transformada:", senha_codificada)
print("Senha recuperada:", senha_recuperada)
```

```python
import hashlib
```

```python
mensagem = input("Digite uma mensagem: ")

hash_gerado = hashlib.sha256(
    mensagem.encode("utf-8")
).hexdigest()

print("Mensagem original:", mensagem)
print("Hash SHA-256:", hash_gerado)
print("Tamanho do hash:", len(hash_gerado))
```

```python
mensagem1 = "senha123"
mensagem2 = "Senha123"

hash1 = hashlib.sha256(mensagem1.encode("utf-8")).hexdigest()
hash2 = hashlib.sha256(mensagem2.encode("utf-8")).hexdigest()

print("Hash da primeira mensagem:", hash1)
print("Hash da segunda mensagem:", hash2)

if hash1 == hash2:
    print("Os hashes são iguais.")
else:
    print("Os hashes são diferentes.")
```

```python
pin_correto = "427"
tentativas = 0
pin_encontrado = False
```

```python
for numero in range(1000):
    tentativa = str(numero).zfill(3)
    tentativas += 1

    if tentativa == pin_correto:
        pin_encontrado = True
        print("PIN encontrado:", tentativa)
        print("Quantidade de tentativas:", tentativas)
        break

if not pin_encontrado:
    print("PIN não encontrado.")
```

```python
senha_correta = "Aula2026"
tentativas = 0
limite = 3

while tentativas < limite:
    senha_digitada = input("Digite a senha: ")

    if senha_digitada == senha_correta:
        print("Acesso autorizado!")
        break

    else:
        tentativas += 1
        restantes = limite - tentativas

        if restantes > 0:
            print("Senha incorreta.")
            print("Tentativas restantes:", restantes)
        else:
            print("Acesso bloqueado após três erros.")
```

```python
senha = input("Crie uma senha para análise: ")

if len(senha) < 8:
    print("A senha é curta. Tente criar uma senha mais longa.")
else:
    print("A senha possui pelo menos 8 caracteres.")

if senha.isdigit():
    print("Atenção: a senha contém apenas números.")

if senha.isalpha():
    print("Atenção: a senha contém apenas letras.")

if len(set(senha)) == 1:
    print("Atenção: a senha repete sempre o mesmo caractere.")

print("Análise concluída.")
```
